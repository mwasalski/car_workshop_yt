```python
%python
CATALOG = 'car_workshop'

PARQUET_BASE     = f'/Volumes/{CATALOG}/bronze/landing'
CHECKPOINT_BASE  = f'/Volumes/{CATALOG}/bronze/landing/autoloader_checkpoints'

print(f'Source :  {PARQUET_BASE}')
print(f'Checkpoints :  {CHECKPOINT_BASE}')
```

```python
%python
# Keeps references to running streams when TRIGGER_MODE = continuous
active_streams = []



def ingest_table(table_name, schema_name='bronze', trigger_mode='availableNow'):
    source_base     = PARQUET_BASE
    checkpoint_base = CHECKPOINT_BASE

    source_path     = f'{source_base}/{table_name}'
    checkpoint_path = f'{checkpoint_base}/{table_name}'
    # schema_location = TABLE_SCHEMAS.get(table_name)
    target_table    = f'{CATALOG}.{schema_name}.{table_name}'
    partition_cols  = PARTITIONED_TABLES.get(table_name)

    print(f'  {table_name}  ->  {target_table}')

    reader = (
        spark.readStream
            .format('cloudFiles')
            .option('cloudFiles.format', 'parquet')
            # .option('cloudFiles.schemaLocation', schema_location)
            .option('cloudFiles.inferColumnTypes', 'true')    # let Auto Loader infer types from parquet
            .option('spark.sql.parquet.datetimeRebaseModeInRead', 'CORRECTED')
            .option('spark.sql.parquet.int96RebaseModeInRead', 'CORRECTED')
    )
    if partition_cols:
        # hive partition columns live in the directory PATH, not inside the parquet
        # files - declare them explicitly instead of relying on schema-inference
        # timing (a stale schemaLocation would otherwise yield NULLs forever)
        reader = reader.option('cloudFiles.partitionColumns', ','.join(partition_cols))
    reader = reader.load(source_path)

    writer = (
        reader.writeStream
            .format('delta')
            .outputMode('append')
            .option('checkpointLocation', checkpoint_path)
            .option('mergeSchema', 'true')
    )

    if partition_cols:
        writer = writer.partitionBy(*partition_cols)

    if trigger_mode == 'availableNow':
        # Process all currently available files, then stop automatically
        query = writer.trigger(availableNow=True).toTable(target_table)
        query.awaitTermination()
        print(f'     done  ({query.lastProgress["numInputRows"] if query.lastProgress else "?"} rows in last micro-batch)')
    else:
        # Keep stream alive; picks up new files every 30 s
        query = writer.trigger(processingTime='30 seconds').toTable(target_table)
        active_streams.append((table_name, query))
        print(f'     stream started  (id={query.id})')

    return query


print('ingest_table() ready')
```

```python
%python
# Loaded directly to Delta by initial_dims.ipynb - listed for validation only
DIM_TABLES = [
    'locations',
    'employees',
    'customers',
    'vehicles',
    'products',
    'services',
    'suppliers',
]


print('=== Dim tables ===')
for table in DIM_TABLES:
    if not SINGLE_TABLE or SINGLE_TABLE == table:
        try:
            ingest_table(table)
        except Exception as e:
            error_msg = str(e)
            if 'CF_EMPTY_DIR_FOR_SCHEMA_INFERENCE' in error_msg or 'empty' in error_msg.lower():
                print(f'     SKIPPED: {table} - source directory is empty')
            elif 'DELTA_METADATA_MISMATCH' in error_msg or 'Partition columns do not match' in error_msg:
                print(f'     ERROR: {table} - partition mismatch (table needs recreation)')
            else:
                print(f'     ERROR: {table} failed - {error_msg[:100]}')
print('Done.')
```
